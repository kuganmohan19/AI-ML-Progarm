```python
import pandas as pd
from sklearn.cluster import KMeans
data = pd.read_csv("kmeans_data.csv")
data.columns = data.columns.str.lower()
x = data[['age', 'income']]
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(x)
clusters = kmeans.predict(x)
data['cluster'] = clusters
print(data)
```
