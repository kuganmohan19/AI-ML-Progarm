```python
graph = {
    'A': [('B', 5), ('C', 1)],
    'B': [('D', 6), ('E', 8)],
    'C': [('F', 2)],
    'D': [],
    'E': [],
    'F': []
}
def hill_climbing(start):
    current = start
    while True:
        neighbors = graph[current]
        if not neighbors:
            break
        next_node = max(neighbors, key=lambda x: x[1])
        print("At node", current, "(value:", next_node[1], ")")
        if next_node[1] <= 0:
            break
        current = next_node[0]
        if not graph[current]:
            print("Reached peak at:", current)
            break
hill_climbing('A')
```
