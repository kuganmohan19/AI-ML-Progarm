```python
import heapq
graph = {
    'A': [('B', 2), ('C', 1)],
    'B': [('D', 5), ('E', 3)],
    'C': [('F', 4)],
    'D': [],
    'E': [],
    'F': []
}
def best_first(start):
    visited = set()
    pq = [(0, start)]
    while pq:
        h, node = heapq.heappop(pq)
        if node not in visited:
            visited.add(node)
            print(node, end=" ")
            for neighbor, heuristic in graph[node]:
                heapq.heappush(pq, (heuristic, neighbor))
best_first('A')
```
