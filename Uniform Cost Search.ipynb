```python
import heapq
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [],
    'E': [],
    'F': []
}
def ucs(start):
    visited = set()
    pq = [(0, start)]
    while pq:
        cost, node = heapq.heappop(pq)
        if node not in visited:
            print("Visited:", node, "Cost:", cost)
            visited.add(node)
            for neighbor, weight in graph[node]:
                heapq.heappush(pq, (cost + weight, neighbor))
ucs('A')
```
