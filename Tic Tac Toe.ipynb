```python
import random
def check_winner(board, player):
    wins = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]
    return any(
        all(board[i] == player for i in combo)
        for combo in wins
    )
def empty_cells(board):
    return [i for i in range(9) if board[i] == ' ']
def ai_move(board):
    for i in empty_cells(board):
        board[i] = 'O'
        if check_winner(board, 'O'):
            board[i] = ' '
            return i
        board[i] = ' '
    for i in empty_cells(board):
        board[i] = 'X'
        if check_winner(board, 'X'):
            board[i] = ' '
            return i
        board[i] = ' '
    if board[4] == ' ':
        return 4
    for i in [0, 2, 6, 8]:
        if board[i] == ' ':
            return i
    return random.choice(empty_cells(board))
board = [' '] * 9
while True:
    print(board[0:3])
    print(board[3:6])
    print(board[6:9])
    move = int(input("Enter position (0-8): "))
    if board[move] != ' ':
        print("Invalid move")
        continue
    board[move] = 'X'
    if check_winner(board, 'X'):
        print("Player Wins")
        break
    if not empty_cells(board):
        print("Draw")
        break
    ai = ai_move(board)
    board[ai] = 'O'
    if check_winner(board, 'O'):
        print("AI Wins")
        break
    if not empty_cells(board):
        print("Draw")
        break
```
